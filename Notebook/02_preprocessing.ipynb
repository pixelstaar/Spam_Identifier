```python
import pandas as pd
import re
```

```python
df = pd.read_csv("../data/raw/indian_spam.csv")

df.columns = ["labels", "messages"]

df["numerical_labels"] = df["labels"].apply(
    lambda x: 0 if x == "ham" else 1
)
```

```python
def clean_text(text):
    text = str(text)
    text = text.lower()
    text = re.sub(r"http\S+|www\S+", " <URL> ", text)
    text = re.sub(r"\d+", "<NUMBER>", text)
    text = re.sub(r"\s+", " ", text).strip()
    return text
```

```python
test_message = "CONGRATULATIONS!!! You won ₹50,000. Visit www.example.com NOW!"

print("Original:", test_message)

print("Cleaned:", clean_text(test_message))
```

```python
df["cleaned_message"] = df["messages"].apply(clean_text)
```
